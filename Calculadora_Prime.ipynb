## COMANDOS QUE APRENDI

*   **PRINT** - mostra a mensagem ou código que quiser, seguido de `()`
*   **INPUT** - solicita uma entrada do usuário, seguido de `()`
*   **EXIT** - encerra a execução do programa, seguido de `()`
*   **FLOAT** - transforma o valor em um número decimal (ponto flutuante)
*   **ELIF** - permite testar uma nova condição se a anterior for falsa ("senão, se...")
*   **ELSE** - executa um bloco de código se todas as condições anteriores forem falsas ("caso contrário")
*   **CHOICE** - seleciona um item aleatório de uma lista (módulo random)
*   **SHUFFLE** - embaralha a ordem dos itens de uma lista (módulo random)
*   **LEN** - fala quantos caractéries tem no seu texto usa `()`
*   **FIND** - acha o caractere que você quiser no lugar mencionado em "lista.find()"
*   **APPEND** - adiciona mais uma palavra ou variável a sua "lista.append()"
*  **INSERT** - escolher a posição da palavra que vc iserirá "lista.insert(2, "palavra")
*  **DEL** - deletar algo, geralmente uma posição em uma lista, "del lista[3]"
